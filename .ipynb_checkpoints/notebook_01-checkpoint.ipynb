15 min:

- Calculate the 90%, 95%, 99% and 99.99% confidence interval for at least 2 columns with numerical data in the dataset that you chose in portfolio assignment 4. Do you see the impact the confidence has on the interval?
- Now calculate the 95% confidence interval again but use only the first 10% of your rows. Compare this interval to the previous 95% confidence interval you calculated. Do you see the impact of having less data?


**Assumption:** Ik denk dat de confidence interval groter wordt naarmate je de zekerheid verhoogt. Ook denk ik dat het confidence interval groter wordt als je een kleinere sample gebruikt.

#### Confidence interval of Financial Loss (in Million $)

```python
confidences = [0.90, 0.95, 0.99, 0.9999]
feature = cybersecurityThreats["Financial Loss (in Million $)"]

for conf in confidences: 
    conf_int = st.t.interval(
        conf,
        len(cybersecurityThreats)-1,
        loc=feature.mean(),
        scale=st.sem(feature)
    )
    print(f"Met {conf}% zekerheid schatten we dat het werkelijke gemiddelde ligt tussen {str(round(conf_int[0], 2))} en {str(round(conf_int[1], 2))}.")
    
```

```python
first_ten_per = cybersecurityThreats.iloc[:int(len(cybersecurityThreats) * 0.1)]
feature = first_ten_per["Financial Loss (in Million $)"]

conf_int = st.t.interval(
    0.95,
    len(first_ten_per)-1,
    loc=feature.mean(),
    scale=st.sem(feature)
)

print(conf_int)
```

#### Confidence interval of Affected Users

```python
confidences = [0.90, 0.95, 0.99, 0.9999]
feature = cybersecurityThreats["Number of Affected Users"]

for conf in confidences: 
    conf_int = st.t.interval(
        conf,
        len(cybersecurityThreats)-1,
        loc=feature.mean(),
        scale=st.sem(feature)
    )
    print(f"Met {conf}% zekerheid schatten we dat het werkelijke gemiddelde ligt tussen {str(round(conf_int[0], 2))} en {str(round(conf_int[1], 2))}.")
    
```

```python
first_ten_per = cybersecurityThreats.iloc[:int(len(cybersecurityThreats) * 0.1)]
feature = first_ten_per["Number of Affected Users"]

conf_int = st.t.interval(
    0.95,
    len(first_ten_per)-1,
    loc=feature.mean(),
    scale=st.sem(feature)
)

print(conf_int)
```

**Findings:** Mijn aanname was correct: naarmate de zekerheid (confidence) toenam, werd het confidence interval inderdaad breder.

```python
contingency_table = pd.crosstab(penguins['sex'], penguins['island'])

chi2 = chi2_contingency(contingency_table)


display(contingency_table)
print(f"P value: {chi2[1]}")
print(f"Odds of correlation: {round((1 -chi2[1])*100, 2)}%")

display(contingency_table.plot(kind='bar'))
```

**Findings:** Er is inderdaad geen sterk bewijs voor een relatie tussen sex en species, en hetzelfde geldt voor sex en island. Mijn aanname bleek dus correct.

30 min: Train a decision tree to predict the species of a penguin based on their characteristics.

- Prepare the data:
    - Note: Some machine learning algorithms can not handle missing values. You will either need to:
        - replace missing values (with the mean or most popular value). For replacing missing values you can use .fillna(<value>) https://pandas.pydata.org/docs/reference/api/pandas.Series.fillna.html
        - remove rows with missing data. You can remove rows with missing data with .dropna() https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.dropna.html
- Split the penguin dataset into a train (70%) and test (30%) set.
- Use the train set to fit a DecisionTreeClassifier. You are free to to choose which columns you want to use as feature variables and you are also free to choose the max_depth of the tree.
- Use your decision tree model to make predictions for both the train and test set.
- Calculate the accuracy for both the train set predictions and test set predictions
Is the accurracy different? Did you expect this difference?
- Use the plot_tree_classification function above to create a plot of the decision tree. Take a few minutes to analyse the decision tree. Do you understand the tree?
- Which depth and features did you add per cycle?
- Optional: Perform the same tasks but try to predict the sex of the pinguin based on the other columns


```python
penguins = sns.load_dataset("penguins")

penguins_clean = penguins.dropna()

features_list = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
target_variable = 'species'

penguins_train, penguins_test = train_test_split(
    penguins_clean,    
    test_size=0.3,
    random_state=42,
    stratify=penguins_clean['species']
)

print(penguins_train.shape, penguins_test.shape)
```

```python
def calculate_accuracy(predictions, actuals):
    if(len(predictions) != len(actuals)):
        raise Exception("The amount of predictions did not equal the amount of actuals")
    
    return (predictions == actuals).sum() / len(actuals)

dt_classification = DecisionTreeClassifier(max_depth = 3)
dt_classification.fit(penguins_train[features_list], penguins_train['species'])

predictionsOnTrainset = dt_classification.predict(penguins_train[features_list])
predictionsOnTestset = dt_classification.predict(penguins_test[features_list])

accuracyTrain = calculate_accuracy(predictionsOnTrainset, penguins_train.species)
accuracyTest = calculate_accuracy(predictionsOnTestset, penguins_test.species)

print("Accuracy on training set " + str(accuracyTrain))
print("Accuracy on test set " + str(accuracyTest))
```

```python
from sklearn import tree
import graphviz

def plot_tree_classification(model, features, class_names):
    dot_data = tree.export_graphviz(model, out_file=None, 
                          feature_names=features,  
                          class_names=class_names,  
                          filled=True, rounded=True,  
                          special_characters=True)  

    # Turn into graph using graphviz
    graph = graphviz.Source(dot_data)  

    # Write out a pdf
    graph.render("decision_tree")

    # Display in the notebook
    return graph 

plot_tree_classification(dt_classification, features_list, np.sort(penguins_clean.species.unique()))
```

**Findings:** Het model presteert goed, wat blijkt uit de hoge en vergelijkbare nauwkeurigheidsscores op zowel de trainings en testset. Dit geeft aan dat het model goed generaliseert en geen last heeft van overfitting. Deze resultaten zijn te verwachten, gezien het aantal en de kwaliteit van de gebruikte features. De gebruikte kenmerken zoals snavellengte, flipperlengte en lichaamsgewicht, zijn waarschijnlijk sterk gecorreleerd met de soort van de pinguïn, waardoor ze effectieve voorspellers zijn in deze classificatietaak.

## Portfolio Assignment 17

30 min: Train a decision tree to predict the body_mass_g of a penguin based on their characteristics.

- Split the penguin dataset into a train (70%) and test (30%) set.
- Use the train set to fit a DecisionTreeRegressor. You are free to to choose which columns you want to use as feature variables and you are also free to choose the max_depth of the tree. Note: Some machine learning algorithms can not handle missing values. You will either need to
    - replace missing values (with the mean or most popular value). For replacing missing values you can use .fillna(<value>) https://pandas.pydata.org/docs/reference/api/pandas.Series.fillna.html
    - remove rows with missing data. You can remove rows with missing data with .dropna() https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.dropna.html
- Use your decision tree model to make predictions for both the train and test set.
- Which depth and features did you add per cycle?
- Calculate the RMSE for both the train set predictions and test set predictions.
- Is the RMSE different? Did you expect this difference?
- Use the plot_tree_regression function above to create a plot of the decision tree. Take a few minutes to analyse the decision tree. Do you understand the tree?


```python
def calculate_rmse(predictions, actuals):
    if(len(predictions) != len(actuals)):
        raise Exception("The amount of predictions did not equal the amount of actuals")
    
    return (((predictions - actuals) ** 2).sum() / len(actuals)) ** (1/2)

def plot_tree_regression(model, features):
    # Generate plot data
    dot_data = tree.export_graphviz(model, out_file=None, 
                          feature_names=features,  
                          filled=True, rounded=True,  
                          special_characters=True)  

    # Turn into graph using graphviz
    graph = graphviz.Source(dot_data)  

    # Write out a pdf
    graph.render("decision_tree")

    # Display in the notebook
    return graph 
    
penguins = sns.load_dataset("penguins")

penguins_clean = penguins.dropna()

features = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm']

X = penguins_clean[features]
y = penguins_clean['body_mass_g']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

dt_regressor = DecisionTreeRegressor(max_depth=3, random_state=42)
dt_regressor.fit(X_train, y_train)

y_train_pred = dt_regressor.predict(X_train)
y_test_pred = dt_regressor.predict(X_test)

train_r2 = calculate_rmse(y_train, y_train_pred)
test_r2 = calculate_rmse(y_test, y_test_pred)

print(f"trainset rmse: {train_r2:.2f}")
print(f"testset rmse: {test_r2:.2f}")

plot_tree_regression(dt_regressor, features)
```

**Findings:** De RMSE op de trainingsset is 335.37 en op de testset 350.41. Deze getallen liggen dicht bij elkaar en zijn niet erg hoog, wat betekent dat het model goed presteert op beide datasets. We merken ook dat hoe meer features we gebruiken, hoe beter het model voorspelt. De RMSE-waarden komen dan dichter bij elkaar te liggen. De gekozen tree depth van 3 werkt goed: als we die diepte veranderen, worden de verschillen tussen de trainingset en testset groter, wat minder goed is.
